```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("CalibratedData_B46_2chains_260105.csv", parse_dates=['DateTime'], delimiter=',', index_col='DateTime', usecols=['DateTime', 'N22_Pressure','N79_Pressure'])

df = df/10 # umrechnung in Pa
```

```python
df.plot()
```

```python
df_slow = df.rolling(window='10min', center=True).mean()
```

```python
df_slow.plot()
plt.title("Slow Pressure Fluctuations in Snow")
```

```python
df_fast = df - df_slow
```

```python
df_fast.plot()
plt.title("Fast Pressure Fluctuations in Snow")
plt.ylabel("(Pa)")
```

```python
df_slow.corr()
```

```python
df_fast.corr()
```

```python
df_fast.describe()
```

```python
df_fast[:100].plot()
```

```python
df_fast_diff = df_fast["N22_Pressure"] - df_fast["N79_Pressure"]
```

```python
df_fast_diff.plot()
plt.title("Difference of Fast Pressure Fluctuations between 0.2 and 0.5 m")
plt.ylabel("(Pa)")
```
